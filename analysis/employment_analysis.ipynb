## 3. Data Cleaning and Preparation

Before creating the visualisations, I need to prepare the dataset to make sure the results are accurate.

During the initial exploration, I found that the dataset contains different occupations, geographic areas and industry classifications. Some occupations also appear in multiple records.

In this section, I will check for missing values, duplicates and possible data quality issues. I will then filter the dataset to focus on the six occupations selected earlier.

The aim is to create a smaller dataset that can be used for comparing employment numbers, salaries and geographic differences.

### 3.1 Checking Missing and Unavailable Values

I checked the employment and salary columns for missing or non-numeric values before preparing the data for visualisation. This helps identify values that could affect calculations and comparisons.

```python

# Columns needed for employment and salary analysis
numeric_columns = ["TOT_EMP", "A_MEAN", "A_MEDIAN", "H_MEAN"]

# Check missing and non-numeric values
for col in numeric_columns:
    numeric_values = pd.to_numeric(df[col], errors="coerce")

    missing_count = df[col].isna().sum()
    invalid_count = (
        df[col].notna() & numeric_values.isna()
    ).sum()

    print(f"{col}:")
    print(f"  Missing values: {missing_count}")
    print(f"  Non-numeric values: {invalid_count}")

    if invalid_count > 0:
        print("  Examples:")
        display(
            df.loc[
                df[col].notna() & numeric_values.isna(),
                col
            ].value_counts().head()
        )

    print()
```

**Findings**

Although Pandas did not identify any missing values in these columns, I found several records containing symbols instead of numbers.

| Column | Non-numeric values |
|---|---:|
| TOT_EMP | 11,598 |
| A_MEAN | 5,352 |
| A_MEDIAN | 7,474 |
| H_MEAN | 20,964 |

These values cannot be used directly in calculations. I will treat them as missing values rather than replacing them with zero, as this could affect the accuracy of the results.

### 3.2 Checking Duplicate Records

I will check the dataset for duplicate rows to see whether any records contain exactly the same information.

Since the dataset includes different occupations, states and industries, some records may look similar without actually being duplicates. This check will help identify any repeated information before preparing the data for visualisation.

```python

# Check for completely identical rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Display duplicate rows if any are found
if duplicate_count > 0:
    display(df[df.duplicated(keep=False)].head(10))
```

**Findings**

The duplicate check returned 0 duplicate rows, meaning no completely identical records were found in the dataset.

Therefore, no rows need to be removed at this stage. However, I still need to consider the different geographic and industry classifications to avoid double-counting employment figures.

### 3.3 Converting Data Types

Some employment and salary columns contain symbols instead of numbers. Before using these columns in calculations or visualisations, I need to convert them into numeric values.

I will create a copy of the original dataset and convert the selected columns using Pandas. Any values that cannot be converted will be treated as missing values rather than zero.

```python

# Create a copy to keep the original dataset unchanged
df_clean = df.copy()

# Convert employment and salary columns to numeric
numeric_columns = ["TOT_EMP", "A_MEAN", "A_MEDIAN", "H_MEAN"]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors="coerce"
    )

# Check the results after conversion
print("Data types after conversion:")
print(df_clean[numeric_columns].dtypes)

print("\nMissing values after conversion:")
display(df_clean[numeric_columns].isna().sum())
```

**Findings**

All four columns were successfully converted to numeric values (`float64`). The special symbols were replaced with `NaN`, allowing Pandas to recognise them as missing values.

The number of missing values matches the non-numeric values found earlier. I also kept the original dataset unchanged by creating a separate copy called `df_clean`.

This means the employment and salary columns are now ready for further analysis, although missing values will still need to be considered when creating visualisations.

### 3.4 Inspecting Possible Outliers

Some occupations may have higher salaries than others, so an unusual value does not necessarily mean the data is incorrect. I will review the salary distribution before deciding whether any values need to be excluded.

The highest salary estimates belong to occupations such as Athletes and Sports Competitors, Chief Executives and Cardiologists, which are outside the focus of this project.

### 3.5 Selecting Relevant Occupations

The original dataset contains many occupations that are not directly related to this project.

I will filter the data using the six occupation codes identified during the initial exploration. Using occupation codes instead of job titles helps keep the selection consistent and avoids including unrelated occupations with similar names.

### 3.6 Preparing National and State-Level Data

The filtered dataset still contains records from different geographic areas and industries.

To avoid double-counting, I will separate national and state-level records and use cross-industry employment estimates. This will allow me to compare the six occupations across the United States and between individual states.

```python

# Check national and state-level classification combinations
classification_check = (
    df_ai[
        df_ai["AREA_TYPE"].isin([1, 2])
    ]
    .groupby(
        ["AREA_TYPE", "NAICS", "I_GROUP", "OWN_CODE"],
        dropna=False
    )
    .size()
    .reset_index(name="Number of Records")
    .sort_values(
        ["AREA_TYPE", "Number of Records"],
        ascending=[True, False]
    )
)

display(classification_check.head(30))
```

The classification check showed that the national cross-industry records contain all six selected occupations.

I also found separate industry-level records, which could lead to double-counting if they were combined with the national totals.

For the analysis, I will use cross-industry records with the appropriate geographic and ownership classifications.

```python

# Check state-level cross-industry records
state_check = df_ai[
    (df_ai["AREA_TYPE"] == 2) &
    (df_ai["NAICS"] == "000000")
]

# Check ownership codes and number of records
display(
    state_check.groupby(
        ["OWN_CODE", "I_GROUP"],
        dropna=False
    ).size().reset_index(name="Number of Records")
)

# Check how many states and occupations are included
print("Number of geographic areas:",
      state_check["AREA_TITLE"].nunique())

print("Number of occupations:",
      state_check["OCC_CODE"].nunique())

display(
    state_check[
        ["AREA_TITLE", "OCC_CODE", "OCC_TITLE", "OWN_CODE"]
    ].head(12)
)
```

## 4. Data Visualisation

### 4.1 National Employment by Occupation (2023)

**Objective**

The aim of this visualisation is to compare the number of people employed in six occupations related to AI, data science and software development across the United States in May 2023.

I want to identify which occupations have the highest and lowest employment levels and understand the differences between them.

I chose a horizontal bar chart because it makes it easier to compare employment figures across different occupations. It also allows longer occupation names to be displayed clearly without overlapping.

```python
from matplotlib.ticker import StrMethodFormatter
# Select the occupation names and employment figures
employment_data = df_national[
    ["OCC_TITLE", "TOT_EMP"]
].sort_values("TOT_EMP", ascending=True)

# Create a horizontal bar chart
plt.figure(figsize=(11, 6))

plt.barh(
    employment_data["OCC_TITLE"],
    employment_data["TOT_EMP"],
    color="steelblue"
)

# Add titles and axis labels
plt.title("National Employment by Occupation (May 2023)")
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Format employment figures on the x-axis
plt.gca().xaxis.set_major_formatter(
    StrMethodFormatter("{x:,.0f}")
)

# Add employment values beside each bar
for index, value in enumerate(employment_data["TOT_EMP"]):
    plt.text(
        value + 15000,
        index,
        f"{value:,.0f}",
        va="center"
    )

plt.xlim(0, employment_data["TOT_EMP"].max() * 1.18)
plt.tight_layout()
plt.show()
```

**Findings**

The chart shows a large difference in employment between the six selected occupations in May 2023.

Software Developers had the highest employment, with approximately 1.66 million workers. This was more than eight times the number of Data Scientists (192,710), who had the second-highest employment.

Computer Programmers had 120,370 workers, followed by Database Architects with 59,920 and Computer and Information Research Scientists with 35,210. Statisticians had the lowest employment, with 29,950 workers.

The results suggest that software development was a much larger occupational field than the other selected areas in 2023.

However, these figures represent total employment within each occupation. They do not indicate how many workers were directly involved in AI-related activities.

### 4.2 Annual Salary Distribution by Occupation (2023)

**Objective**

The aim of this visualisation is to compare the distribution of average annual salaries for the six selected occupations across US states in May 2023.

I want to investigate which occupations generally offer higher salaries and whether salaries vary significantly between states.

I chose a box plot because it shows more than just the average salary. It allows me to compare the median, spread and possible outliers for each occupation.

This should help me understand whether salary levels are consistent across states or if there are large differences depending on the location.

```python
# Select occupation names and mean annual salaries
salary_data = df_states[
    ["OCC_TITLE", "A_MEAN"]
].dropna(subset=["A_MEAN"]).copy()

# Check the number of available salary records
salary_data.groupby("OCC_TITLE")["A_MEAN"].count()
```

```python
plt.figure(figsize=(12, 7))

sns.boxplot(
    data=salary_data,
    x="A_MEAN",
    y="OCC_TITLE",
    color="steelblue"
)

plt.title("Distribution of Mean Annual Wages Across States (May 2023)")
plt.xlabel("Mean Annual Wage (USD)")
plt.ylabel("Occupation")

# Format salary values on the x-axis
from matplotlib.ticker import StrMethodFormatter

plt.gca().xaxis.set_major_formatter(
    StrMethodFormatter("${x:,.0f}")
)

plt.tight_layout()
plt.show()
```

**Findings**

The box plot shows differences in mean annual wages across US states for the six selected occupations in May 2023.

Computer and Information Research Scientists had the highest median of state-level mean wages, at approximately $130,000. Database Architects and Software Developers also showed relatively high salary levels.

Computer Programmers and Statisticians had lower median state-level mean wages, at around 95,000 to $100,000.

The chart also shows differences between states within the same occupation. Some occupations have a wider salary range, while several have potential outliers. For example, Computer and Information Research Scientists had a particularly high value above $200,000.

One limitation is that not every occupation has salary data for all 51 geographic areas. Computer and Information Research Scientists had only 42 available records.

Overall, the results suggest that both occupation and location are important when comparing salaries. However, these figures represent state-level mean wages rather than individual workers' salaries.

### 4.3 Employment by US State and Occupation (2023)

**Objective**

The aim of this visualisation is to investigate how employment in the six selected occupations varies across US states in May 2023.

I want to identify which states have the highest employment levels and whether the same states have high employment across different occupations.

I chose a heatmap because it allows me to compare several occupations and states in one visualisation. The colours make it easier to identify patterns and differences without having to compare a large number of individual bars.

To keep the chart readable, I will focus on the ten states with the highest combined employment across the selected occupations.

```python
# Select state, occupation and employment data
state_employment = df_states[
    ["AREA_TITLE", "OCC_TITLE", "TOT_EMP"]
].copy()

# Identify the ten states with the highest combined employment
top_states = (
    state_employment.groupby("AREA_TITLE")["TOT_EMP"]
    .sum(min_count=1)
    .nlargest(10)
    .index
)

# Keep only the top ten states
top_state_employment = state_employment[
    state_employment["AREA_TITLE"].isin(top_states)
]

# Prepare the data for the heatmap
employment_heatmap = top_state_employment.pivot(
    index="AREA_TITLE",
    columns="OCC_TITLE",
    values="TOT_EMP"
)

# Order states by their combined employment
employment_heatmap = employment_heatmap.loc[top_states]

employment_heatmap
```

```python
plt.figure(figsize=(15, 8))

sns.heatmap(
    employment_heatmap,
    cmap="Blues",
    annot=True,
    fmt=",.0f",
    linewidths=0.5,
    cbar_kws={"label": "Estimated Employment"}
)

plt.title("Employment by State and Occupation (May 2023)")
plt.xlabel("Occupation")
plt.ylabel("State")

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

**Findings**

The heatmap shows how employment in the six selected occupations was distributed across the ten states with the highest combined employment in May 2023.

California had the highest combined employment among these states. It also had the highest number of Software Developers, with 304,390 workers, followed by Texas with 138,510 and New York with 105,460.

Software Developers had the highest employment figures across all ten states, which explains why this column has the darkest colours in the heatmap.

California also had the highest number of Data Scientists among the states shown, with 33,220 workers. This suggests that employment in several technology-related occupations is particularly concentrated in California.

One limitation is that the large Software Developer figures dominate the colour scale, making smaller differences between other occupations harder to identify. Some employment values are also missing, such as Computer and Information Research Scientists in Massachusetts.

Overall, the results suggest that employment in these selected occupations is not evenly distributed across US states. However, the chart shows total employment figures rather than employment relative to each state's population or workforce size.

### 4.4 Relationship Between Employment and Annual Salary (2023)

**Objective**

The aim of this visualisation is to investigate whether there is a relationship between employment levels and average annual salaries across US states for the six selected occupations.

I want to see whether states with higher employment also tend to have higher salaries, or whether the relationship varies depending on the occupation.

I chose a scatter plot because it is useful for exploring the relationship between two numerical variables. In this case, I will compare total employment and mean annual wages.

Each point represents one occupation in one state. Using different colours for the occupations will help me identify patterns and compare the results.

```python
# Select the columns needed for the scatter plot
scatter_data = df_states[
    ["AREA_TITLE", "OCC_TITLE", "TOT_EMP", "A_MEAN"]
].copy()

# Remove records without employment or salary values
scatter_data = scatter_data.dropna(
    subset=["TOT_EMP", "A_MEAN"]
)

# Check how many records are available for each occupation
scatter_data.groupby("OCC_TITLE").size()
```

```python
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=scatter_data,
    x="TOT_EMP",
    y="A_MEAN",
    hue="OCC_TITLE",
    alpha=0.7,
    s=70
)
plt.xscale("log")
plt.title("Employment vs Mean Annual Wage Across US States (May 2023)")
plt.xlabel("Total Employment (Logarithmic Scale)")
plt.ylabel("Mean Annual Wage (USD)")

# Format the axes
from matplotlib.ticker import LogLocator, FuncFormatter

plt.gca().xaxis.set_major_locator(
    LogLocator(base=10, subs=(1, 2, 5))
)

plt.gca().xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x:,.0f}")
)
plt.gca().yaxis.set_major_formatter(
    StrMethodFormatter("${x:,.0f}")
)

plt.legend(
    title="Occupation",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

**Findings**

The scatter plot compares employment levels and mean annual wages across US states for the six selected occupations in May 2023.

The results show that Software Developers generally have higher employment figures than the other occupations. However, states with higher employment do not always have higher salaries.

Computer and Information Research Scientists have some of the highest mean annual wages, including values above $180,000, despite having relatively low employment figures compared with Software Developers.

The original chart had most points grouped on the left because of the large differences in employment. I used a logarithmic scale to make the smaller values easier to compare.

Overall, there is no clear positive relationship between employment and mean annual wages across all six occupations. This suggests that employment size alone is not enough to explain salary differences.

However, the chart does not establish whether employment directly affects salaries. Other factors, such as location and occupation, may also influence wages.
