## 2. Identifying AI-Related Occupations

The dataset contains employment information for many different jobs across the United States. Since this project focuses on AI, Data Science and Software Development, I first need to identify which occupations are relevant to the analysis.

To do this, I will search for job titles such as Data Scientist, Software Developer and Computer Programmer. I will also use occupation codes to help identify the correct records.

One thing to consider is that not everyone working in these occupations necessarily works with AI. For this reason, the selected jobs will be considered AI-related occupations rather than an exact representation of AI employment.

```python
keywords = [
    "Data Scientist",
    "Software Developer",
    "Computer and Information Research Scientist",
    "Computer Programmer",
    "Database Architect",
    "Statistician"
]

matching_occupations = df[
    df["OCC_TITLE"].str.contains(
        "|".join(keywords),
        case=False,
        na=False
    )
]

display(
    matching_occupations[
        ["OCC_CODE", "OCC_TITLE"]
    ].drop_duplicates().sort_values("OCC_TITLE")
)
```

### 2.1 Occupation Selection and Data Considerations

The occupation search uses job titles and SOC codes to identify occupations relevant to the project.

The selected occupations represent different areas of technology, including software development, data analysis, computing research and statistical work. However, these occupations should be treated as AI-related rather than a direct measurement of the entire AI workforce.

The dataset also includes different geographic and industry classifications. These must be considered when selecting records for visualisation to avoid double-counting employment estimates.

### 2.2 Findings from the Initial Exploration

After searching the dataset, I identified six occupations relevant to AI, Data Science and Software Development. These include Data Scientists, Software Developers, Computer Programmers, Database Architects, Statisticians and Computer and Information Research Scientists.

I noticed that some occupations appeared with more than one occupation code. To keep the analysis consistent, I will use the detailed occupation codes rather than the broader categories.

I also found that the dataset contains employment information at both national and state levels, as well as records from different industries.

For the next stage, I will focus on cross-industry employment data and use the appropriate geographic records depending on what I want to compare. This should help avoid counting employment figures more than once.

```python
selected_codes = [
    "15-1251",  # Computer Programmers
    "15-1221",  # Computer Research Scientists
    "15-2051",  # Data Scientists
    "15-1243",  # Database Architects
    "15-1252",  # Software Developers
    "15-2041"   # Statisticians
]

selected_data = df[
    df["OCC_CODE"].isin(selected_codes)
]

display(
    selected_data[
        ["AREA_TYPE", "NAICS", "I_GROUP", "OWN_CODE"]
    ].drop_duplicates().head(30)
)
```
